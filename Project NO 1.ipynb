```python
def rule_based_chatbot():
    print("Chatbot: Hello! I am a rule-based AI. Type 'exit' or 'bye' to stop talking to me.\n")

    while True:

        user_input = input("You: ").lower().strip()

        # Rule 1: Exit Commands
        if user_input in ['exit', 'bye', 'quit', 'goodbye']:
            print("Chatbot: Goodbye! Have a great day.")
            break

        # Rule 2: Greetings
        elif user_input in ['hi', 'hello', 'hey', 'hola']:
            print("Chatbot: Hello there! How can I help you today?")

        # Rule 3: Identifying the bot
        elif "who are you" in user_input or "your name" in user_input:
            print("Chatbot: I am a simple Rule-Based AI created for Project 1.")

        # Rule 4: Well being
        elif "how are you" in user_input:
            print("Chatbot: I'm functioning within normal parameters! How about you?")

        # Rule 5: Capabilities
        elif "what can you do" in user_input:
            print("Chatbot: I can answer basic greetings, tell you who I am, and run in this loop until you say 'bye'.")

        # Default fallback (if no rules match)
        else:
            print("Chatbot: I'm sorry, I don't understand that command yet. Try saying 'hello' or 'who are you'.")

rule_based_chatbot()
```
